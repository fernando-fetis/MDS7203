# Byte pair encoding

```python
def BPE(text, k):

    # Dividir el texto en palabras y luego en caracteres:
    words = text.strip().split()
    corpus = [[c for c in word] for word in words]
    vocab = set(c for word in corpus for c in word)

    # Aplicar merges hasta alcanzar tamaño k de vocabulario:
    while len(vocab) < k:

        # Contar todas las apariciones de bigramas:
        stats = {}
        for word in corpus:
            for i in range(len(word) - 1):
                pair = (word[i], word[i+1])
                stats[pair] = stats.get(pair, 0) + 1

        if not stats:
            break

        # Elegir el bigrama más frecuente:
        best = max(stats, key=stats.get)
        a, b = best
        ab = a + b

        # Hacer el merge del bigrama más frecuente en todo el corpus:
        new_corpus = []
        for word in corpus:
            new_word = []
            i = 0
            while i < len(word):
                if i < len(word)-1 and word[i] == a and word[i+1] == b:
                    new_word.append(ab)
                    i += 2
                else:
                    new_word.append(word[i])
                    i += 1
            new_corpus.append(new_word)

        corpus = new_corpus
        vocab.add(ab)

    return corpus
```

```python
texto = (
    "cantar canta cantando cantan canté "
    "comer come comiendo comen comí "
    "vivir vive viviendo viven viví "
    "caminar camina caminando caminé "
    "amar ama amando amé aman "
    "romper rompe rompiendo rompió rompieron"
)

tokens = BPE(texto, k=20)

for w, t in zip(texto.split(), tokens):
    print(f"{w} -> {' '.join(t)}")
```

# Sinusoidal embedding

```python
import numpy as np
import matplotlib.pyplot as plt

seq_length = 100
emb_dim = 512

token_embeddings = np.random.randn(seq_length, emb_dim)

def positional_encoding(T, D):
    pos = np.arange(T)[:, np.newaxis]  # [T, 1].
    i = np.arange(D // 2)[np.newaxis, :]  # [1, D/2].
    angle_rates = 1 / np.power(10000, (2 * i) / D)  # [1, D/2].
    angle_rads = pos * angle_rates  # [T, D/2].
    pe = np.zeros((T, D))
    pe[:, 0::2] = np.sin(angle_rads)
    pe[:, 1::2] = np.cos(angle_rads)
    return pe  # [T, D].

positional_encodings = positional_encoding(seq_length, emb_dim)
combined = token_embeddings + positional_encodings

# Gráfico:
fig, axs = plt.subplots(3, 1, figsize=(10, 8))

axs[0].imshow(token_embeddings)
axs[0].set_title('Token embedding')

axs[1].imshow(positional_encodings)
axs[1].set_title('Positional encoding')

axs[2].imshow(combined)
axs[2].set_title('Token embedding + positional encoding')

for ax in axs:
    ax.set_xlabel('dimensión de embedding')
    ax.set_ylabel('Posición en la secuencia')

plt.tight_layout()
plt.show()
```
