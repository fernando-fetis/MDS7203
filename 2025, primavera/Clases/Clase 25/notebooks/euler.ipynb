# Algoritmo de Euler

```python
from matplotlib import pyplot as plt
import torch
```

```python
class Euler:

    def __init__(self, f):
        self.f = f

    def simulate(self, x0, T, dt=0.01):

        t = 0
        x_list = [x0]
        
        while t < T:
            x = x_list[-1] + dt * self.f(x_list[-1], t)
            x_list.append(x)
            t += dt
            
        return x_list
    
    def plot_trajectory(self, x0_list, T):

        plt.figure(figsize=(10,5))
        for x0 in x0_list:
            x = self.simulate(x0, T)
            t = torch.linspace(0,T, len(x))
            plt.plot(t, x, label=rf'$x_0$={x0}')

        plt.xlabel(r'$t$')
        plt.ylabel(r'$x_t$')
        plt.legend()
        plt.grid(alpha=0.2)
        plt.show()
```

```python
def f(x, t):
    return x * torch.sin(x) - x.exp() + t

euler = Euler(f)

x0_list = torch.linspace(-1, 1, 5)
euler.plot_trajectory(x0_list, T=2)
```

```python
def f(x, t):
    return -2 * x

euler = Euler(f)

x0_list = torch.linspace(-10, 10, 5)
euler.plot_trajectory(x0_list, T=3)
```
